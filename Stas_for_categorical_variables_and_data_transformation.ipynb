```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Bestsellers with categories (1).csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Genre'].value_counts()
```

```python
df['Genre_code'] = df['Genre'].astype('category').cat.codes
df['Genre_code'].median()
```

```python
genre_counts = df['Genre'].value_counts()

plt.figure(figsize=(8,5))
genre_counts.plot(kind='bar', color='skyblue')
plt.title('Bar Chart of Genres')
plt.xlabel('Genre')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(6,6))
genre_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title('Pie Chart of Genres')
plt.ylabel('')
plt.show()
```

```python
num_df = df.select_dtypes(include='number')

print(num_df.head())

plt.figure(figsize=(10,6))
num_df.boxplot()
plt.title('Box Plot of Numerical Features')
plt.xticks(rotation=45)
plt.show()
```

```python
num_df_norm = (num_df - num_df.min()) / (num_df.max() - num_df.min())

print(num_df_norm.head())

import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
num_df_norm.boxplot()
plt.title('Box Plot of Normalized Numerical Features')
plt.xticks(rotation=45)
plt.show()
```
