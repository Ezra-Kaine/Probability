```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Bestsellers with categories (1).csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    print(f"\nColumn: {col}")
    print(df[col].value_counts())
```

```python
bins = [0, 3, 7, 10]
labels = ['Poor', 'Good', 'Excellent']

df['Rating_category'] = pd.cut(df['User Rating'], bins=bins, labels=labels, include_lowest=True)

df[['User Rating', 'Rating_category']].head()
```

```python
num_df = df.select_dtypes(include='number')

skewness = num_df.skew()

print("Skewness of numerical features:\n", skewness)
```

```python
rating_genre = df.groupby('Genre')['User Rating'].mean()
print(rating_genre)
```

```python
correlation = df['User Rating'].corr(df['Price'])
print(f"Correlation between User Rating and Price: {correlation}")
```
